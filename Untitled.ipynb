```python
import pandas as pd

# 1. 读入 clean 后的队伍级数据
df = pd.read_csv("lol_team_clean.csv")

# 2. 去掉关键列缺失的行
sub = df.dropna(subset=["golddiffat10", "win"]).copy()

# 3. 定义金钱差分箱
bins = [-10000, -3000, -2000, -1000, 0, 1000, 2000, 3000, 10000]
labels = [
    "< -3000",
    "-3000 ~ -2000",
    "-2000 ~ -1000",
    "-1000 ~ 0",
    "0 ~ 1000",
    "1000 ~ 2000",
    "2000 ~ 3000",
    "> 3000"
]

sub["gold_bin"] = pd.cut(sub["golddiffat10"], bins=bins, labels=labels)

# 4. 按分箱计算胜率 & 样本数
gold_win = (
    sub.dropna(subset=["gold_bin"])
       .groupby("gold_bin", observed=False)["win"]
       .agg(win_rate="mean", count="size")
       .reset_index()
)

# 5. 把胜率保留三位小数，方便前端使用
gold_win["win_rate"] = gold_win["win_rate"].round(3)

gold_win
```
